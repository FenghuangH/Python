```python
# Control flow
labels = []
for t in temps:
    if t >= 18:
        labels.append("warm")
    else:
        labels.append("cool")
print(labels)
```

```python
labels = []
i = 0
while i < len(temps):
    t = temps[i]
    if t >= 18:
        labels.append("warm")
    else:
        labels.append("cool")
    i += 1
print(labels)
```

```python
# List comprehension (equivalent)
labels2 = ["warm" if t>=18 else "cool" for t in temps]
print(labels2)
```


# 4) Matplotlib — quick charts & customization


```python
plt.figure(figsize=(10,4)) # (width, height)
plt.plot(df['date'], df['rentals'], linewidth=1) # x, y
plt.title('Daily Rentals') # title
plt.xlabel('Date'); plt.ylabel('rentals') # axis labels
plt.grid(True, alpha=0.3) # grid
plt.show() # show the plot
```

```python
df.groupby('weekday_name')['rentals'].mean()
```

```python
# Bar chart from a groupby
avg_by_day = df.groupby('weekday_name')['rentals'].mean().reindex(
    ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'] 
) 
plt.figure(figsize=(8,4))
avg_by_day.plot(kind='bar')
plt.title('Average Rentals by Weekday')
plt.ylabel('rentals')
plt.show()
```

```python

fig, ax = plt.subplots(1,2, figsize=(12,4)) #Two subplots side by side, 1 row, 2 columns
ax[0].plot(df['date'], df['rentals'])
ax[0].set_title('Daily Rentals')
ax[0].set_ylabel('rentals')

ax[1].scatter(df['temp_c'], df['rentals'], s=10, alpha=0.6)
ax[1].set_title('Rentals vs Temperature')
ax[1].set_xlabel('temp_c')
ax[1].set_ylabel('rentals')
plt.show()
```

```python
# Interactivity
try:
    from ipywidgets import interact, interactive, widgets, fixed
    _WIDGETS_OK = True
except Exception as e:
    print("ipywidgets not available; interactive parts will be minimal. Error:", e)
    _WIDGETS_OK = False

    
if _WIDGETS_OK:
    @interact(min_temp=(-5,35,1), max_temp=(0,40,1), rain=widgets.Dropdown(options=['All','No rain','Rain'], value='All'))
    def filter_plot(min_temp=10, max_temp=30, rain='All'):
        q = (df['temp_c']>=min_temp) & (df['temp_c']<=max_temp)
        if rain == 'No rain':
            q &= (df['is_rain']==0)
        elif rain == 'Rain':
            q &= (df['is_rain']==1)
        dff = df[q]
        plt.figure(figsize=(6,4))
        sns.scatterplot(data=dff, x='temp_c', y='rentals', hue='marketing', alpha=0.7)
        plt.title(f'Scatter (n={len(dff)})')
        plt.show()
else:
    print("Widgets unavailable; skipping interactive demo.")
```
